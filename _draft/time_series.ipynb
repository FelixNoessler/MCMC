### Stationarity

- strong statitionarity:  
$(y_{t_1} ... y_{t_n})'$ and $ (y_{t_1 + h} ... y_{t_n + h})'$ have the same underlying distribution

- weak stationarity:  
$(y_{t_1} ... y_{t_n})'$ and $ (y_{t_1 + h} ... y_{t_n + h})'$ have identical first and second moments;   
this implies that the $E(y_t) = u$ and $Var(y_t) = v$ are constant and $Cov(y_t, y_s)$ depends only on $|t-s|$ 


### Autocovariance
- general definition:  
$\gamma(t,s)=Cov(y_t, y_s) = E[(y_t-u_s)(y_s-u_s)]$, with $E(y_t)=u_t$ and $E(y_s)=u_s$

- when there is stationarity, $E(y_t)=u$ for all $t$:  
$\gamma(t,s) = \gamma(|t-s|)$
lag $h$ $\gamma(h)=Cov(y_t, y_{t-h})$

### Autocorrelation (ACF)
- general definition:  
$\rho(t,s)=\frac{\gamma(t,s)}{\sqrt{\gamma(t,t)} \cdot \sqrt{\gamma(s,s)}}$, $\gamma(t,t)$ variance at time point $t$, $\sqrt{\gamma(t,t)}$ standard deviation at time point t
- for stationarity:  
$\rho(h)=\frac{\gamma(h)}{\gamma(0)}$, $\gamma(0) = Var(y_t)$

### Sample Autocovariance

- for stationarity and time series $y_{1:T}$:  
$\hat \gamma(h) = \frac{1}{T}\sum_{t=1}^{T-h} (y_{t+h} - \bar y)(y_t - \bar y)$, with $\bar y = \frac{1}{T}\sum_{t=1}^{T}y_t$


### Sample Autocorrelation
- for stationarity and time series $y_{1:T}$:  
$\hat \rho(h) = \frac{\hat \gamma(h)}{\hat \gamma(0)}$

```r


white_noise = rnorm(100, 0, 1)

data_table = tibble(white_noise, t=1:100)
acf_table = tibble(
    lag=0:10, 
    acf=acf(white_noise, lag.max=10, plot=F)$acf)

pacf_table = tibble(
    lag=1:10, 
    pacf=pacf(white_noise, lag.max=10, plot=F)$acf)


ggplot()+
    geom_line(aes(t, white_noise), data=data_table)+
    theme_classic()

ggplot()+
    geom_point(aes(lag, acf), data=acf_table)+
    theme_classic()

ggplot()+
    geom_point(aes(lag, pacf), data=pacf_table)+
    theme_classic()
```
